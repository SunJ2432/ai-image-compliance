```python
from google.colab import userdata
def optional_secret(name):
    try:
        return userdata.get(name)
    except Exception:
        return None
hf_token = optional_secret("HF_TOKEN")
kaggle_token = optional_secret("KAGGLE_API_TOKEN")
kaggle_username = optional_secret("KAGGLE_USERNAME")
print("HF token configured:", bool(hf_token))
print("Kaggle API token configured:", bool(kaggle_token))
print("Kaggle username configured:", bool(kaggle_username))
if hf_token: os.environ["HF_TOKEN"] = hf_token
if kaggle_token: os.environ["KAGGLE_API_TOKEN"] = kaggle_token
if kaggle_username: os.environ["KAGGLE_USERNAME"] = kaggle_username
```
